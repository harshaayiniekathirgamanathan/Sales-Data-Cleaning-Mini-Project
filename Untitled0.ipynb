```python
cleaned = df.copy()
```

```python
text_columns = [
    'CustomerName',
    'City',
    'Product',
    'OrderStatus'
]

for col in text_columns:

    cleaned[col] = cleaned[col].astype('string').str.strip()
```

```python
cleaned['CustomerName'] = cleaned['CustomerName'].str.title()

cleaned['City'] = cleaned['City'].str.title()

cleaned['Product'] = cleaned['Product'].str.title()

cleaned['OrderStatus'] = cleaned['OrderStatus'].str.title()
```

```python
for col in text_columns:

    cleaned[col] = cleaned[col].replace("", pd.NA)
```

```python
cleaned[['CustomerName', 'City', 'Product', 'OrderStatus']].head(10)
```

```python
before = len(cleaned)

cleaned = cleaned.drop_duplicates()

after = len(cleaned)

print("Duplicates removed:", before - after)

print("Remaining rows:", len(cleaned))
```

```python
quantity_text = cleaned['Quantity'].astype('string').str.strip().str.lower()

quantity_text = quantity_text.replace({
    'three': '3',
    '2 pcs': '2'
})
```

```python
cleaned['Quantity'] = pd.to_numeric(
    quantity_text,
    errors='coerce'
)
```

```python
cleaned.loc[
    (cleaned['Quantity'] <= 0) |
    (cleaned['Quantity'] > 100),
    'Quantity'
] = np.nan
```

```python
price_text = (
    cleaned['UnitPrice']
    .astype('string')
    .str.strip()
    .str.replace(',', '', regex=False)
)
```

```python
cleaned['UnitPrice'] = pd.to_numeric(
    price_text,
    errors='coerce'
)
```

```python
cleaned.loc[
    (cleaned['UnitPrice'] <= 0) |
    (cleaned['UnitPrice'] > 100000),
    'UnitPrice'
] = np.nan
```

```python
age_text = (
    cleaned['CustomerAge']
    .astype('string')
    .str.strip()
    .str.lower()
)

age_text = age_text.str.replace(
    ' years', '', regex=False
)

age_text = age_text.str.replace(
    ' year', '', regex=False
)

age_text = age_text.replace({
    'forty': '40'
})
```

```python
cleaned['CustomerAge'] = pd.to_numeric(
    age_text,
    errors='coerce'
)
```

```python
cleaned.loc[
    (cleaned['CustomerAge'] < 18) |
    (cleaned['CustomerAge'] > 100),
    'CustomerAge'
] = np.nan
```

```python
numeric_columns = [
    'Quantity',
    'UnitPrice',
    'CustomerAge'
]

for col in numeric_columns:

    cleaned[col] = cleaned[col].fillna(
        cleaned[col].median()
    )
```

```python
categorical_columns = [
    'CustomerName',
    'City',
    'Product',
    'OrderStatus'
]

for col in categorical_columns:

    cleaned[col] = cleaned[col].fillna(
        cleaned[col].mode()[0]
    )
```

```python
cleaned.isnull().sum()
```

```python
cleaned['OrderDate'] = pd.to_datetime(
    cleaned['OrderDate'],
    errors='coerce',
    format='mixed',
    dayfirst=True
)
```

```python
print(cleaned['OrderDate'].isnull().sum())
```

```python
valid_dates = cleaned['OrderDate'].dropna().sort_values()

median_date = valid_dates.iloc[len(valid_dates) // 2]

cleaned['OrderDate'] = cleaned['OrderDate'].fillna(
    median_date
)
```

```python
cleaned['OrderDate'].isnull().sum()
```

```python
cleaned['OrderID'] = cleaned['OrderID'].astype('Int64')

cleaned['Quantity'] = cleaned['Quantity'].round().astype('Int64')

cleaned['CustomerAge'] = cleaned['CustomerAge'].round().astype('Int64')

cleaned['UnitPrice'] = cleaned['UnitPrice'].astype(float)

cleaned['OrderDate'] = pd.to_datetime(cleaned['OrderDate'])
```

```python
cleaned.dtypes
```

```python
cleaned['TotalAmount'] = (
    cleaned['Quantity'] * cleaned['UnitPrice']
)
```
